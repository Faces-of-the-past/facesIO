# Importing data from RKD

In this example we'll show how to use `Collection` to import data from a RKD `xml` file.

```python
from facesIO.classes import Collection
SOURCE_RKD="../data/rkd_example.xml"
```

```python
collection = Collection(SOURCE_RKD, adapter="RKD")
```

A `Collection` is a list of artworks. Let's take a look at them (notice we use the `str` function for turning each `Artwork` object in a human-readable string):

```python
[str(artwork) for artwork in collection.artworks]
```

If we just open one of them, we'll find much more:

```python
single_artwork = collection.artworks[0]
single_artwork
```

To extract some basic field, such as `title` or `artists`, we just do:

```python
single_artwork.title
```
