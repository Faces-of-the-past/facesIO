# Importing data from RKD

In this example we'll show how to use `Collection` to import data from a RKD `xml` file.

```python
from facesIO.classes import Collection
SOURCE_RKD="../data/rkd_example.xml"
```

```python
collection = Collection.from_source(SOURCE_RKD, museum="RKD")
```

A `Collection` is a list of artworks. Let's take a look at them (notice we use the `str` function for turning each `Artwork` object in a human-readable string):

```python
collection
```

```python
[str(artwork) for artwork in collection]
```

If we just open one of them, we'll find much more:

```python
single_artwork = collection.artworks[0]
single_artwork
```

To extract some basic field, such as `title` or `artist`, we just do:

```python
single_artwork.title
```

```python
single_artwork.as_jsonl_line()
```

## Import directly from list of `Artwork`s

```python
artworks = collection.artworks
identical_collection = Collection(artworks)
identical_collection
```

## Export to `jsonl` (and reimport)

```python
collection.as_jsonl("test.jsonl")
```

```python
reimported_collection = Collection.from_jsonl("test.jsonl")
```

```python
[str(artwork) for artwork in reimported_collection]
```

## Combine collections
`Collections` can be added to each other

```python
bigger_collection = collection + identical_collection
bigger_collection
```
